```python
import pandas as pd
import matplotlib.pyplot as plt
```

Import the monthly order and revenue data


```python
monthly_data = {
    "month": [
        "2016-09", "2016-10", "2016-12",
        "2017-01", "2017-02", "2017-03", "2017-04",
        "2017-05", "2017-06", "2017-07", "2017-08",
        "2017-09", "2017-10", "2017-11", "2017-12",
        "2018-01", "2018-02", "2018-03", "2018-04",
        "2018-05", "2018-06", "2018-07", "2018-08",
        "2018-09"
    ],
    "orders": [
        3, 308, 1,
        789, 1733, 2641, 2391,
        3660, 3217, 3969, 4293,
        4243, 4568, 7451, 5624,
        7220, 6694, 7188, 6934,
        6853, 6160, 6273, 6452,
        1
    ],
    "revenue": [
        267.36, 49507.66, 10.90,
        120312.87, 247303.02, 374344.30, 359927.23,
        506071.14, 433038.60, 498031.48, 573971.68,
        624401.69, 664219.43, 1010271.37, 743914.17,
        950030.36, 844178.71, 983213.44, 996647.75,
        996517.68, 865124.31, 895507.22, 854686.33,
        145.00
    ]
}

df = pd.DataFrame(monthly_data)

df
```

Visualise the monthly revenue

```python
plt.figure(figsize=(12, 6))

plt.plot(df["month"], df["revenue"])

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (R$)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Then focus on 2017-2018 period
Visualise the revenue and the order

```python
df_main = df[
    (df["month"] >= "2017-01") &
    (df["month"] <= "2018-08")
].copy()

df_main
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df_main["month"], df_main["revenue"])

plt.title("Monthly Revenue (2017–2018)")
plt.xlabel("Month")
plt.ylabel("Revenue (R$)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("monthly_revenue.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df_main["month"], df_main["orders"])

plt.title("Monthly Orders (2017–2018)")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("monthly_orders.png", dpi=300, bbox_inches="tight")

plt.show()
```
